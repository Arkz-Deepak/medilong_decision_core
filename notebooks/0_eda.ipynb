# Definition of Gaurdrail

```python
def check_safety_guardrails(row):
    # Fix for decision_core: Convert 1-row DataFrame to a Series
    if isinstance(row, pd.DataFrame):
        row = row.iloc[0]
        
    overrides = []
    
    # Note: If a value is NaN, NaN > 4.5 evaluates to False (which safely skips the override)
    if row.get('TSH (mIU/L)', 0) > 4.5:
        overrides.append("OVERRIDE: TSH > 4.5 mIU/L (Suspected Hypothyroidism)")
    
    if row.get('PRL(ng/mL)', 0) > 25.0:
        overrides.append("OVERRIDE: PRL > 25.0 ng/mL (Suspected Hyperprolactinemia)")
        
    return {
        "safety_override": len(overrides) > 0,
        "reasons": overrides
    }
```

# Creting a Shap Tree Explainer

```python
explainer = shap.TreeExplainer(base_xgb)

sample = X_test.iloc[[0]]
shap_value = explainer(sample)

print("Base value (starting log-odds):", shap_value.base_values[0])
print("Per-feature values:", shap_value.values[0])
print("Feature names:", sample.columns.tolist())
```

# Defining Decision Core Model of System

```python
def decision_core(patient_record, explainer, calibrated_model):

    # 1. Run Guardrails 
    safety_override = check_safety_guardrails(patient_record)
    
    # 2. Predict Probability (Extracts the probability of class 1)
    risk_prob = calibrated_model.predict_proba(patient_record)[0, 1]
    
    # 3. Categorize Risk
    if risk_prob < 0.30:
        risk_category = "Low"
    elif risk_prob <= 0.65:
        risk_category = "Moderate"
    else:
        risk_category = "High"
        
    # 4. Extract SHAP Contributions
    shap_vals = explainer(patient_record)
    
    # Zip together the feature names, raw data values, and SHAP impact values
    contributions = [
        {"feature": feat, "value": val, "contribution": float(s_val)}
        for feat, val, s_val in zip(patient_record.columns, shap_vals.data[0], shap_vals.values[0])
    ]
    
    # Sort to isolate the highest positive and lowest negative contributing features
    positive_contribs = sorted(
        [c for c in contributions if c["contribution"] > 0], 
        key=lambda x: x["contribution"], 
        reverse=True
    )
    negative_contribs = sorted(
        [c for c in contributions if c["contribution"] < 0], 
        key=lambda x: x["contribution"]
    )
    
    # Extract the top 2 positive and top 2 negative feature
    top_shap = []
    if positive_contribs:
        top_shap.append(positive_contribs[:2])
    if negative_contribs:
        top_shap.append(negative_contribs[:2])
        
    # 5. Construct Output Payload
    payload = {
        "risk_probability": float(risk_prob),
        "risk_category": risk_category,
        "safety_override": safety_override,
        "shap_contributions": top_shap
    }
    
    return payload
```

```python
sample_patient = X_test.iloc[[0]]
result_1 = decision_core(sample_patient, explainer, calibrated_model)
print("--- Test Case 1: Standard Inference ---")
print(result_1)
```
